### file handeling

File Handling ka use files ko create, read, write, update, aur delete karne ke liye kiya jata hai.
| Mode | Description                                                 |
| ---- | ----------------------------------------------------------- |
| `r`  | File ko read karne ke liye (default mode)                   |
| `w`  | File me write karne ke liye, purana data delete ho jata hai |
| `a`  | File ke end me data add (append) karne ke liye              |
| `x`  | Nayi file create karne ke liye                              |
| `r+` | Read aur write dono                                         |
| `w+` | Read aur write, purana data delete ho jata hai              |
| `a+` | Read aur append dono                                        |
| `rb` | Binary file read karne ke liye                              |
| `wb` | Binary file write karne ke liye                             |


Common Functions Table
| Function       | Description                          |
| -------------- | ------------------------------------ |
| `open()`       | File open karta hai                  |
| `read()`       | Puri file read karta hai             |
| `readline()`   | Ek line read karta hai               |
| `readlines()`  | Sabhi lines list me return karta hai |
| `write()`      | File me data likhta hai              |
| `writelines()` | Multiple lines likhta hai            |
| `close()`      | File close karta hai                 |

property
| Property               | Description                                                                              |
| ---------------------- | ---------------------------------------------------------------------------------------- |
| **Data Persistence**   | Data permanently store hota hai, program band hone ke baad bhi data save rehta hai.      |
| **Read Operation**     | File se data read kar sakte hain.                                                        |
| **Write Operation**    | File me naya data likh sakte hain.                                                       |
| **Append Operation**   | Existing file ke end me data add kar sakte hain bina purana data delete kiye.            |
| **File Creation**      | Nayi file create kar sakte hain.                                                         |
| **File Deletion**      | Zarurat padne par file delete kar sakte hain.                                            |
| **Random Access**      | `seek()` aur `tell()` functions ki madad se kisi bhi position par access kar sakte hain. |
| **Data Sharing**       | Files ko multiple users/programs ke beech share kiya ja sakta hai.                       |
| **Security**           | File permissions ke through data ko secure rakha ja sakta hai.                           |
| **Large Data Storage** | Bahut bada data efficiently store kiya ja sakta hai.                                     |

```python
class InvalidAgeError(Exception):
    pass
try:
    age=15
    if age < 18:
        raise InvalidAgeError("you are not eligible to vote")
except InvalidAgeError as e:
    print(e)
```

```python
class insufficientFundsError(Exception):
    pass
try:
    balance=1000
    withdrawal_amount=1500
    if withdrawal_amount > balance:
        raise insufficientFundsError("you have insufficient funds to make this withdrawal")
    print("hello world")
except insufficientFundsError as e:
    print(e)
```

```python
class InvalidMobileNumberError(Exception):
    pass
try:
    mobile_number="12345"
    if len(mobile_number) != 10:
        raise invalidmobileNumberError("invalid mobile number")
except invalidmobileNumberError as e:
    print(e)
```

```python
class EmailAlreadyExistsError(Exception):
    pass
try:
    email="shraddha@gmail.com"
    if email == "shraddha@gmail.com":
        raise EmailAlreadyExistsError("email already exists")
except EmailAlreadyExistsError as e:
    print(e)
```
