### Numpy
NumPy is a Python library used for numerical and mathematical operations. It provides support for fast multidimensional arrays and various mathematical functions.

**Types of NumPy Arrays**

NumPy mainly has these types of arrays:

| Type of Array | Dimension       | Description                       | Example           |
| ------------- | --------------- | --------------------------------- | ----------------- |
| 1D Array      | One Dimension   | Contains elements in a single row | `[1, 2, 3, 4]`    |
| 2D Array      | Two Dimension   | Contains rows and columns         | `[[1,2],[3,4]]`   |
| 3D Array      | Three Dimension | Collection of 2D arrays           | `[[[1,2],[3,4]]]` |

```python
# ==============================
# NumPy Array Operations Example
# ==============================

import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 6, 8])

print("Array 1 :", arr1)
print("Array 2 :", arr2)

print("\n========== OPERATIONS ==========\n")

# Multiplication
print("1️⃣ Multiplication (*)")
print("arr1 * arr2 =", arr1 * arr2)

# Division
print("\n2️⃣ Division (/)")
print("arr1 / arr2 =", arr1 / arr2)

# Addition
print("\n3️⃣ Addition (+)")
print("arr1 + arr2 =", arr1 + arr2)

# Subtraction
print("\n4️⃣ Subtraction (-)")
print("arr1 - arr2 =", arr1 - arr2)

# Positive Sign with Multiplication
print("\n5️⃣ Multiplication with Unary Plus (* +)")
print("arr1 * +arr2 =", arr1 * +arr2)

# Floor Division
print("\n6️⃣ Floor Division (//)")
print("arr1 // +arr2 =", arr1 // +arr2)

print("\n✅ All operations completed successfully!")
```
