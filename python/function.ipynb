```python
#prime number 1 to 100
def prime():
    for x in range(1,101):
        if x>1:
            for i in range (2,x):
                if x%i ==0:
                    break
            else:
                print(x)
                    
prime()
```

```python
def prime(lst):

    for x in lst:

        if x > 1:

            for i in range(2, x):

                if x % i == 0:
                    print(x, "is not prime")
                    break

            else:
                print(x, "is prime")


numbers = [2, 4, 5, 7, 9]

prime(numbers)
```
