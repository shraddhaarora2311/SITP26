### Encapsulation

***Encapsulation** is one of the important concepts of Object-Oriented Programming (OOP) in Python.
It means wrapping data (variables) and methods (functions) together inside a class and restricting direct access to some data.

It helps in:

Data hiding
Security
Better control over data
Preventing accidental modification
Real Life Example

Think of an ATM Machine:

You can use ATM functions like withdraw and deposit.
But you cannot directly access the bank server data.

This is called Encapsulation.
