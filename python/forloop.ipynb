2. while loop

### while loop in python
A while loop is used to repeat a block of code again and again as long as a condition is True.

It is called an entry-controlled loop because the condition is checked before executing the loop body.


```python
#while condition
#code of blocks:



x = 1#staring point
while x<=5:
    print(x)
    x+=1
```

```python
x = 5
while x>=1 :
    print(x)
    x-=1
```

```python
x=1
while x<=10:
    if x ==5:
        break
    print(x)
    x+=1
```
