### Basics of Inheritance in Python

Inheritance is an important concept of Object-Oriented Programming (OOP) in Python.
It allows one class to reuse the properties and methods of another class.

The class whose properties are inherited is called Parent Class / Base Class / Super Class
The class that inherits properties is called Child Class / Derived Class / Sub Class


| Term              | Meaning                              |
| ----------------- | ------------------------------------ |
| Parent Class      | Class whose properties are inherited |
| Child Class       | Class that inherits another class    |
| Reusability       | Reusing existing code                |
| Method Overriding | Child class changes parent method    |

| Type                     | Meaning                                    |
| ------------------------ | ------------------------------------------ |
| Single Inheritance       | One child inherits one parent              |
| Multiple Inheritance     | One child inherits multiple parents        |
| Multilevel Inheritance   | Child inherits from another child class    |
| Hierarchical Inheritance | Multiple child classes inherit one parent  |
| Hybrid Inheritance       | Combination of different inheritance types |





* Inheritance is a feature of OOP (Object-Oriented Programming) where one class can use the properties and methods of another class.

* Parent Class / Base Class → The class whose properties are inherited.
* Child Class / Derived Class → The class that inherits properties.

        Inheritance helps in:

* Code reusability
* Reducing duplicate code
* Easy maintenance


Types of Inheritance in Python


| Type                     | Meaning                                    |
| ------------------------ | ------------------------------------------ |
| Single Inheritance       | One parent → One child                     |
| Multiple Inheritance     | Multiple parents → One child               |
| Multilevel Inheritance   | Grandparent → Parent → Child               |
| Hierarchical Inheritance | One parent → Multiple children             |
| Hybrid Inheritance       | Combination of different inheritance types |

```python
#5 prime number

class PrimeNumberChecker:
    def check_prime(self, num):

        if num > 1:
            for i in range(2, num):
                if num % i == 0:
                    print(num, "is Not Prime Number")
                    break
            else:
                print(num, "is Prime Number")

        else:
            print(num, "is Not Prime Number")
            
obj = PrimeNumberChecker()
obj.check_prime(9)
```

```python
#11 bank account
class BankFunction:
    balance = 0
    def deposit(self):
        amount=int(input("enter the amount:"))
        self.balance = self.balance + amount
        print("Total Balance:", self.balance)

    def withdraw(self):
        amount = int(input("Enter Withdraw Amount: "))

        if amount <= self.balance:
            self.balance = self.balance - amount
            print("Remaining Balance:", self.balance)

        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Current Balance:", self.balance)
obj = BankFunction()
obj.deposit()
obj.withdraw()
obj.check_balance()
```
