```python
#pie chart
a=[1,2,3,4]
label =["apple","banana","cherry","date"]
plt.pie(a,labels=label)
plt.show()
```

```python
a=[1,2,3,4]
label =["apple","banana","cherry","date"]
plt.pie(a,labels=label, explode=[0.1,0.2,0,0])
plt.show()
```

```python
a=[1,2,3,4]
label =["apple","banana","cherry","date"]
plt.pie(a,labels=label, explode=[0,0.2,0,0],shadow=True)
plt.show()
```

```python
a=[1,2,3,4]
label =["apple","banana","cherry","date"]
colors=["purple","black","yellow","pink"]
plt.pie(a,labels=label, explode=[0,0.2,0,0],shadow=True, colors=colors)
plt.show()
```

```python
a=[1,2,3,4]
label =["apple","banana","cherry","date"]
colors=["purple","black","yellow","pink"]
plt.pie(a,labels=label, explode=[0,0.2,0,0],shadow=True, colors=colors,autopct='%1.1f%%')
plt.legend()
plt.show()
```
